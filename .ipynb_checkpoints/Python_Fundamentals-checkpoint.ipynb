Formatting Output

```python
pi = 3.14159
r = 2.5
h = 10**0.5
volume = pi * r**2 * h
print(
    'A cylinder of radius',
    r,
    'and height',
    h,
    'has volume',
    volume
)
out_string = (
    'A cylinder of radius {0:.2f} and height {1:.2f} '
    'has volume {2:.2f}'
).format(r, h, volume)

print(out_string)
```

## Activity 2: Functions, Lists, Tuples and Loops

```python
import math
def area_circle(radius):
    return math.pi * radius**2

def area_cylinder(radius, height):
    circle_area = area_circle(radius)
    side_area = 2 * radius * math.pi * height
    return 2 * circle_area + side_area

print('The area of a circle of radius 1 is', round(area_circle(1), 2))
r = 2
height = 10
print('The surface area of a cylinder with radius', r)
print(
    'and height',
    height,
    'is',
    round(area_cylinder(r, height), 2)
)
```

### Task 1

```python
def rectangle_area(length, width):
    return length * width

def rectangular_solid_surface_area(length, width, height):
    front_back = rectangle_area(length, height)
    top_bottom = rectangle_area(length, width)
    left_right = rectangle_area(width, height)

    return 2 * (front_back + top_bottom + left_right)
```

```python
rectangular_solid_surface_area(2, 3, 4)
```

### Task 3

```python
L = [2, 21, 12, 8, 5, 31]

i = 0

while i < len(L):
    j = i + 1 # Only compares each pair once
    while j < len(L):
        print(L[i], L[j])
        j += 1
    i += 1
```
